```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
yellow_1 = pd.read_parquet(r"dataset\yellow_tripdata_2026-01.parquet")
yellow_1.info()
```

```python
green_1 = pd.read_parquet(r"dataset\green_tripdata_2026-01.parquet")
green_1.info()
```

```python
yellow_col = yellow_1.columns
green_col = green_1.columns

common_col = yellow_col.intersection(green_col)
diff_col = yellow_col.difference(green_col)

print(common_col)
print(diff_col)
```
